Muffinplot makes ensembles named "date.str_params_filename.dat##" but that is not an informative file name. This code renames ensembles by renaming 'date' with some root that you'd prefer to call it. 

```python
import os


# path to directory
path = r"C:/Users/prowa/OneDrive - Syracuse University/cGenie/Pubs/userconfigs/Devon370Ma_050xRemin_1xpCO2/f"

filenames = os.listdir(path)

for filename in filenames:
    old_path = os.path.join(path, filename)
    new_path = os.path.join(path, filename.replace("250930", "Devon370Ma_050xRemin_1xpCO2")) # enter the file root (date) and what you want to chnge it to
    print("Renaming:", old_path, "->", new_path)  # debug
    os.rename(old_path, new_path)
```
