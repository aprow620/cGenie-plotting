```python
import pandas as pd
import shutil
import os


transitions = pd.read_csv("C:/Users/prowa/cGENIE_processing/Results/Results_Jan/valid_transitions_anytempchange.csv")
filenames_bkg = transitions['file_bg'].to_list()
filenames_ev = transitions['file_ev'].to_list()

# print(filenames_bkg)
save_dir = r"D:/Transitions/transition" # destination

root_dir = r"D:/Final Runs"  #origin

# full_path_bkg = {}

tar_index = {}
for root, _, files in os.walk(root_dir):
    for f in files:
        if f.endswith(".tar.gz"):
            tar_index[f]=os.path.join(root,f)
print("First Background") 

for index, filename in enumerate(filenames_bkg):
    tar_name = filename + ".tar.gz"

    if tar_name not in tar_index:
        print(f"File not found: {tar_name} for Transition {index+1}")
        continue

    src = tar_index[tar_name]

    new_dir = f"{save_dir}_{index+1}"
    sub_new_dir = os.path.join(new_dir, "background")
    os.makedirs(sub_new_dir, exist_ok=True)
    try:
        shutil.copy(src, sub_new_dir)
    except FileNotFoundError:
         print(f"File not found: {full_path_bkg[index]}")

print ("Now Kellwasser")

for index, filename in enumerate(filenames_ev):
    new_dir = f"{save_dir}_{index+1}"
    sub_new_dir = os.path.join(new_dir, "kellwasser")
    os.makedirs(sub_new_dir, exist_ok=True)

    # Case 1: filename is already a directory
    if os.path.isdir(filename):
        try:
            shutil.copytree(
                filename,
                os.path.join(sub_new_dir, os.path.basename(filename)),
                dirs_exist_ok=True
            )
        except FileNotFoundError:
            print(f"Directory not found: {filename} for Transition {index+1}")
        continue

    # Case 2: filename should correspond to a tar.gz
    tar_name = filename + ".tar.gz"

    if tar_name not in tar_index:
        print(f"File not found: {tar_name} for Transition {index+1}")
        continue

    src = tar_index[tar_name]

    try:
        shutil.copy(src, sub_new_dir)
    except FileNotFoundError:
        print(f"File not found: {src} for Transition {index+1}")
```

```python
def untar_all_to_temp(from_folder, to_folder="temp"):
    os.makedirs(to_folder, exist_ok=True)  # Create temp directory if needed

    all_items = glob.glob(os.path.join(from_folder, "*"))  # Grab everything in from_folder

    for item_path in all_items:
        name = os.path.basename(item_path) #extracts last component of given path

        # Case 1: item is a file
        if os.path.isfile(item_path):
            if tarfile.is_tarfile(item_path):
                print(f"Extracting {name} to {to_folder}")
                try:
                    with tarfile.open(item_path, 'r:*') as tar:
                        tar.extractall(path=to_folder)
                except Exception as e:
                    print(f"Failed to extract {name}: {e}")
            else:
                print(f"Copying file {name} to {to_folder}")
                try:
                    shutil.copy2(item_path, to_folder)
                except Exception as e:
                    print(f"Failed to copy file {name}: {e}")

        # Case 2: item is a directory (not a tar file)
        elif os.path.isdir(item_path):
            print(f"Copying contents of directory {name} to {to_folder}")
            try:
                for root, dirs, files in os.walk(item_path):
                    # relative path inside the directory
                    rel_path = os.path.relpath(root, from_folder)
                    dest_dir = os.path.join(to_folder, rel_path)
        
                    os.makedirs(dest_dir, exist_ok=True)
        
                    for file in files:
                        src_file = os.path.join(root, file)
                        dst_file = os.path.join(dest_dir, file)
                        shutil.copy2(src_file, dst_file)
            except Exception as e:
                print(f"Failed to copy directory {name}: {e}")
```
